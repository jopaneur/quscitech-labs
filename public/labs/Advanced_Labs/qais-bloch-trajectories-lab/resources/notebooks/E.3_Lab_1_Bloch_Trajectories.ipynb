**Task 1 - Helper Utilities**

```python
# ---- Figure helper (robust; standardized for all labs) ----
import os, matplotlib.pyplot as plt

def save_e_figure(fig_label: str,
                  fname: str,
                  subdir: str = "Advanced_Labs/figures",
                  fig=None, ax=None):
    """
    Save the current/explicit figure with a consistent IEEE-compliant path.
    - Does NOT inject figure numbers/titles into the visual.
    - Appends .png if missing.
    - Uses bbox_inches='tight' to avoid clipping.
    """
    os.makedirs(subdir, exist_ok=True)

    if fig is None:
        fig = plt.gcf()
    if ax is None:
        ax = fig.axes[0] if fig.axes else None
    if ax is None:
        print("⚠️ No axes found. Draw a plot first, or pass fig/ax explicitly.")
        return

    if not fname.lower().endswith(".png"):
        fname += ".png"

    outpath = os.path.join(subdir, fname)
    fig.tight_layout()
    fig.savefig(outpath, dpi=160, bbox_inches="tight")

    print(f"Saved figure → {outpath}\n{fig_label}")
```

**Methodology Analysis**

This helper defines a standardized function to save all figures with IEEE Appendix E–compliant filenames. It ensures consistency by creating the target directory if missing, appending “.png” when not specified, and applying tight_layout() to prevent clipping. The helper accepts either the current active figure or an explicit figure/axis handle, then saves the plot at 160 dpi. Finally, it prints a confirmation with the full file path and figure label, providing traceability for Appendix E cross-referencing.

**Task 3 - Hadamard on |0⟩ → measurement counts — Figure E.3.1a**

This module establishes the simplest possible Bloch-sphere trajectory and measurement test: applying a Hadamard gate to the initial state ∣0⟩. The Hadamard creates the equal superposition (∣0⟩ + ∣1⟩)/√2, so repeated measurement produces a balanced histogram with roughly 50% probability for ∣0⟩ and 50% for ∣1⟩. This example introduces the statistical nature of quantum measurement and provides the baseline reference for later multi-qubit and parameterized circuits.

```python
# --- Hadamard on |0> ---
qc = QuantumCircuit(1, 1)
qc.h(0)
qc.measure(0, 0)

res = backend.run(transpile(qc, backend), shots=2000).result()
counts = res.get_counts()

# Plot
plot_histogram(counts, title="H on |0⟩ → ~50/50 outcomes")
fig, ax = plt.gcf(), plt.gca()

# Save
save_e_figure("Figure E.3.1a", "P2_AdvLab01_E.3.1a.png",
              subdir="Advanced_Labs/figures", fig=fig, ax=ax)
plt.show()
```

**Figure E.3.1a — Hadamard on ∣0⟩: Measurement Distribution**

Applying a Hadamard gate to the initial state ∣0⟩ creates the equal superposition (∣0⟩ + ∣1⟩)/√2. Over many measurement shots, this produces a histogram with approximately 50% probability in ∣0⟩ and 50% in ∣1⟩, showing the statistical balance of superposition.

**Expected Results**

The histogram should display nearly equal bar heights for ∣0⟩ and ∣1⟩. Small deviations from exact 50/50 split are expected due to finite sampling.

**Technical Analysis (for the visual)**

The Hadamard gate rotates the Bloch vector from the north pole (∣0⟩) to the equator of the Bloch sphere. Measurement in the computational basis projects the state onto Z, yielding ∣0⟩ or ∣1⟩ with equal probability. The plot is a statistical confirmation of the superposition principle.

**Intuition Sidebar**

Think of the Hadamard as flipping a fair coin in quantum form. Each run gives either “heads” (∣0⟩) or “tails” (∣1⟩), but the overall frequencies converge to 50/50 when repeated many times.

**Task 4 - Compose parameterized block + entangler — Figure E.3.1b**

```python
# --- Parameterized rotation block and entangler ---
theta = Parameter("θ")
block = QuantumCircuit(1, name="R_block")
block.ry(theta, 0)

# Bind parameter and build a 2-qubit circuit
ans = QuantumCircuit(2, 2)
bound_block = block.assign_parameters({theta: np.pi/3})
ans.append(bound_block.to_instruction(), [0])
ans.h(1)
ans.cx(0, 1)
ans.measure([0, 1], [0, 1])

res2 = backend.run(transpile(ans, backend), shots=2000).result()
counts2 = res2.get_counts()

# Plot
plot_histogram(counts2, title="Parameterized block + entangler (θ = π/3)")
fig, ax = plt.gcf(), plt.gca()

# Save
save_e_figure("Figure E.3.1b", "P2_AdvLab01_E.3.1b.png",
              subdir="Advanced_Labs/figures", fig=fig, ax=ax)
plt.show()
```

**Figure E.2.1b — Parameterized Block + Entangler (θ = π/3)**

A single-qubit rotation RY(θ) on the control qubit, followed by a Hadamard on the target and a controlled-X (CX), produces a two-qubit entangled state whose measurement statistics depend smoothly on θ. This module shows how a local parameter and an entangler jointly shape the distribution over computational basis outcomes.

**Expected Results**

With θ = π⁄3, the histogram should be skewed toward states with control qubit = 0. In particular, ∣00⟩ and ∣01⟩ should have higher, roughly equal bars, while ∣10⟩ and ∣11⟩ should be lower and roughly equal. (Numerically, cos²(θ⁄2)=cos²(π⁄6)=0.75 and sin²(θ⁄2)=0.25, so the total weight splits ≈75% on q₀=0 and ≈25% on q₀=1, with each pair divided about evenly by the prior Hadamard on the target.)

**Technical Analysis (for the visual)**

Starting from ∣00⟩, applying RY(θ) on the control yields
cos(θ⁄2)∣0⟩ + sin(θ⁄2)∣1⟩.

On the matrix level, parameterized gates such as RY, RX, and RZ are just 2×2 unitaries with tunable trigonometric entries. For example:

* RY(θ) = [[cos(θ/2), −sin(θ/2)], [sin(θ/2), cos(θ/2)]]

* RX(ϕ) = [[cos(ϕ/2), −i·sin(ϕ/2)], [−i·sin(ϕ/2), cos(ϕ/2)]]

* RZ(λ) = diag(e^(−iλ/2), e^(iλ/2))

Composing RX followed by RZ multiplies their matrices, showing how single-qubit “knobs” build rotations in different planes of the Bloch sphere. In this experiment, we use a single RY block to tilt the control qubit’s amplitude away from the pole by an angle θ.

* A Hadamard on the target creates (∣0⟩+∣1⟩)/√2, so before CX the amplitudes for ∣00⟩ and ∣01⟩ are each cos(θ⁄2)/√2, and for ∣10⟩ and ∣11⟩ are each sin(θ⁄2)/√2.

* The CX gate acts only when the control is ∣1⟩, flipping the target. This swaps the target bits for ∣10⟩ and ∣11⟩ but preserves their equal probabilities.

Therefore the final probability distribution is:

* Pr(q₀=0) = cos²(θ⁄2), split evenly between ∣00⟩ and ∣01⟩.

* Pr(q₀=1) = sin²(θ⁄2), split evenly between ∣10⟩ and ∣11⟩.

* At θ = π⁄3, this matches the expected ≈0.375 bars for ∣00⟩ and ∣01⟩, and ≈0.125 for ∣10⟩ and ∣11⟩.

**Intuition Sidebar**

Treat RY(θ) as the knob that tilts how often the control is ∣0⟩ vs ∣1⟩; the Hadamard spreads that choice across the target, and CX is the gear that couples them—only when the control is ∣1⟩ does the target flip. The histogram is the classical shadow of this coupled behavior: more tilt toward ∣0⟩ on the control means more mass on ∣00⟩/∣01⟩.

**Technical Spotlight — Parameterized Gate Matrices**

Quantum rotation gates are unitary matrices with entries determined by trigonometric functions. For instance:

* RX(ϕ) = [[cos(ϕ/2), −i·sin(ϕ/2)], [−i·sin(ϕ/2), cos(ϕ/2)]]

* RZ(λ) = diag(e^(−iλ/2), e^(iλ/2))

* RY(θ) = [[cos(θ/2), −sin(θ/2)], [sin(θ/2), cos(θ/2)]]

These gates can be composed to build richer single-qubit operations. For example, applying an RX(π/4) followed by an RZ(π/6) produces a combined unitary:

from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
import numpy as np

u = QuantumCircuit(1)
u.rx(np.pi/4, 0)
u.rz(np.pi/6, 0)

print("Unitary on single qubit:\n", np.array(Operator(u)))

Sample output (values truncated):

Unitary on single qubit:
[[0.9239-0.1913j  -0.1913-0.3062j]
 [-0.1913-0.3062j  0.9239+0.1913j]]


This demonstrates that parameterized circuits are not just “black boxes” — they correspond to explicit unitary matrices that encode continuous rotations on the Bloch sphere. By combining parameterized gates (RX, RY, RZ) and entangling gates (CX), QAIS pipelines sculpt probability distributions in a controlled, designable way.
