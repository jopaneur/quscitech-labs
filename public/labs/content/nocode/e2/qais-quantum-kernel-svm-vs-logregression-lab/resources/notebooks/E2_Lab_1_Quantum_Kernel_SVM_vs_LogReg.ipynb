**Task 1 - Figure Helper Utilities**

Centralized utilities for **Appendix E** compliance. No plots generated here.

**Description (what it does):**

* Defines save_e_figure(...) that saves exactly one file per plot with IEEE-style names

* Uses a single figures root for all labs: Beginner_Labs/figures/

* Enforces Beginner Part 1 naming: P1_BeginLab01_E.1.<index><sub>.png (e.g., P1_BeginLab01_E.1.1a.png)

* Provides label_subplots(...) for (a)–(d) subfigure tags

```python
# ---- Figure helper (robust; use in every coded lab) ----
import os, matplotlib.pyplot as plt

def save_e_figure(fig_label: str,
                  fname: str,
                  subdir: str = "Intermediate_Labs/figures",
                  fig=None, ax=None):
    """Save the current/explicit figure with a prefixed label and consistent path."""
    os.makedirs(subdir, exist_ok=True)
    if fig is None:
        fig = plt.gcf()
    if ax is None:
        ax = fig.axes[0] if fig.axes else None
    if ax is None:
        print("⚠️ No axes found. Draw a plot first, or pass fig/ax explicitly.")
        return
    title = ax.get_title() or ""
    if not title.startswith(fig_label):
        ax.set_title((fig_label + " — " + title).strip(" —"))
    outpath = os.path.join(subdir, fname)
    fig.tight_layout()
    fig.savefig(outpath, dpi=160)
    print("Saved", outpath)
```

**Methodology Analysis**

We standardize figure saving and labeling via a helper that stamps the IEEE figure label into the title and writes images into a consistent Beginner_Labs/figures path.

The helper centralizes Appendix-compliant naming with `E_PART="E.1"`, `LAB_STEM="P1_BeginLab01"`, and an output root `Beginner_Labs/figures"`.  
Each `save_e_figure(fig_index, sub)` writes an image uniquely keyed to its E.1 label.  
Ensure you execute Task 1 before saving figures — successful saves print confirmations such as:  
`Saved figure → Beginner_Labs/figures/P1_BeginLab01_E.1.1a.png`.

**Task 3: Comparing Quantum and Classical Classifiers**

Here, the experiment contrasts how a quantum kernel-based support vector machine (SVM) and a classical logistic regression model separate data. The logistic regression decision boundary, visualized as a straight dividing line, illustrates how a linear classifier partitions the dataset. Quantum kernels, by contrast, build similarity measures in high-dimensional Hilbert space, which can capture non-linear patterns. The process flow is: encode data into quantum states, compute kernel overlaps, train the SVM, and then visualize classification performance. The plot highlights the baseline — a simple linear split — against which quantum performance is measured. This matters because it emphasizes where quantum models could reveal richer structures not visible to purely linear classical approaches.


**Exected Results**

*Kernel Heatmap (Quantum Kernel Similarity)*

The heatmap should display bright diagonal entries, since each input is perfectly similar to itself.

Off-diagonal entries will vary, creating structured banding or clustering patterns.

Inputs that are close in the original feature space will appear brighter (high overlap), while distant inputs will appear darker (low overlap).

The overall structure reflects the nonlinear geometry that quantum kernels induce, often sharper and richer than classical kernels.

**Quantum Kernel Similarity Heatmap**

*Figure E.2.1a — Quantum Kernel Similarity Heatmap.*  
The heatmap displays pairwise fidelities of quantum-encoded states, where brighter values indicate greater similarity. Structured diagonal bands reveal clusters of inputs that the quantum kernel perceives as nearly identical, supporting its role as a nonlinear similarity measure for classification.


**Methodology**
Participants compute a quantum kernel matrix by encoding training data into quantum states and evaluating their pairwise inner products (fidelities). The kernel captures the similarity structure of the dataset in Hilbert space. The resulting matrix is plotted as a heatmap, where brighter cells represent stronger similarity between two inputs.

**Technical Analysis**

The visual demonstrates the expressive power of quantum feature maps.

Diagonal dominance confirms normalization: each state overlaps fully with itself.

Clustered bands reveal that quantum encoding can group similar samples, an essential ingredient for classification.

The heatmap directly encodes the Gram matrix of inner products, which serves as the foundation for kernel-based learning algorithms like Support Vector Machines (SVMs).

In Quantum AI contexts, this ability to reshape similarity space is what enables quantum-enhanced classification performance compared with linear models.

**Intuition Sidebar — “Seeing Quantum Geometry”**

Think of the heatmap like a social network map: each person (data point) is compared with every other. Brighter cells mean two “friends” are alike, darker ones mean they’re strangers. The quantum kernel changes the rules of similarity, highlighting patterns that a classical eye would miss — it’s like putting on quantum glasses to see hidden connections.

**Intuition Sidebar — “Drawing a Straight Line”**

But if they’re tangled in curves or circles, the ruler can’t bend to separate them. The quantum kernel, in contrast, can “bend” space itself so that the separation becomes easy.
