**Task 4 - Stress Test Simulation**

This code block stress-tests two contrasting quantum–AI pipelines under realistic gate noise and readout loss.

*QALIS-style model (Pipeline A)* means a small, noise-aware design inspired by Quantum AI Layered Integrated Systems. Concretely, we use an entangling feature map and a simple mitigation step: linear zero-noise extrapolation (ZNE) that estimates what features would look like at near-zero noise.

*CRQC–LLM baseline (Pipeline B)* represents a Conventional Random Quantum Circuit approach paired conceptually with a classical LLM interpreter. In practice here, it is shallower and unmitigated: local rotations only, no entangler, no mitigation.

We inject single-qubit depolarizing noise into common gates and simulate lossy readout by probabilistically flipping the least significant bit of measured bitstrings. As adversarial noise or loss increases, the QALIS pipeline should retain higher scores than CRQC–LLM for a broad range of settings. The robustness gap — the accuracy difference QALIS minus CRQC–LLM — widens with stronger loss channels and moderate noise, then may taper when both systems saturate near failure.

```python
# This module uses `AerSimulator` with a fixed random seed (`seed_simulator=777`) to ensure reproducible noisy runs.
# === Stress-test primitives for Figures E.2.7a–b ===
# Requirements upstream: numpy as np; fmap_A, fmap_B; dataset X, y; readout params w, b

from typing import Dict
from collections import Counter
from qiskit_aer import AerSimulator
from qiskit_aer.noise import NoiseModel, depolarizing_error
from qiskit import transpile

# Reproducible CPU simulator
backend = AerSimulator(seed_simulator=777)

def _noise_model_dep1(p: float) -> NoiseModel:
    """1-qubit depolarizing noise attached to common single-qubit gates."""
    nm = NoiseModel()
    nm.add_all_qubit_quantum_error(depolarizing_error(p, 1), ["id", "rz", "sx", "x"])
    return nm

def _apply_readout_flip(cnt: Dict[str, int], flip_prob: float) -> Dict[str, int]:
    """Flip the LSB with probability flip_prob, keep counts non-negative."""
    c2 = Counter(cnt)
    for bitstring, v in list(cnt.items()):
        if v <= 0:
            continue
        flips = int(round(flip_prob * v))
        if flips <= 0:
            continue
        c2[bitstring] -= flips
        flipped = bitstring[:-1] + ("1" if bitstring[-1] == "0" else "0")
        c2[flipped] += flips
    for k in list(c2.keys()):
        if c2[k] < 0:
            c2[k] = 0
    return dict(c2)

def counts_with_noise(fmap, x, shots: int = 1000, p: float = 0.02, ro_flip: float = 0.03) -> Dict[str, int]:
    """Run fmap(x) with depolarizing noise p and LSB readout flips ro_flip, return counts."""
    nm = _noise_model_dep1(p)
    qc = fmap(x).copy()
    qc.measure_all()
    tqc = transpile(qc, backend)
    result = backend.run(tqc, shots=shots, noise_model=nm).result()
    return _apply_readout_flip(result.get_counts(), ro_flip)

def _features_from_counts(cnt: Dict[str, int], shots: int) -> np.ndarray:
    """Return [⟨Z⊗I⟩, ⟨I⊗Z⟩] from 2-qubit bitstring counts."""
    p00 = cnt.get("00", 0) / shots
    p01 = cnt.get("01", 0) / shots
    p10 = cnt.get("10", 0) / shots
    p11 = cnt.get("11", 0) / shots
    mz0 = (p00 + p01) - (p10 + p11)  # Z⊗I
    mz1 = (p00 + p10) - (p01 + p11)  # I⊗Z
    return np.array([mz0, mz1], dtype=float)

def mitigated_pred(fmap, X: np.ndarray, w: np.ndarray, b: float,
                   shots: int = 1000, p: float = 0.02, ro_flip: float = 0.03, scale: float = 1.5) -> np.ndarray:
    """Zero-noise extrapolation: f0 ≈ 2 f(p) − f(scale·p), then classify with w·f + b."""
    yhat = []
    for x in X:
        c1 = counts_with_noise(fmap, x, shots=shots, p=p, ro_flip=ro_flip)
        c2 = counts_with_noise(fmap, x, shots=shots, p=scale * p, ro_flip=ro_flip)
        f0 = 2.0 * _features_from_counts(c1, shots) - _features_from_counts(c2, shots)
        yhat.append(1 if (w @ f0 + b) >= 0 else 0)
    return np.array(yhat, dtype=int)

def unmitigated_pred(fmap, X: np.ndarray, w: np.ndarray, b: float,
                     shots: int = 1000, p: float = 0.02, ro_flip: float = 0.03) -> np.ndarray:
    """No mitigation: compute f(p) at finite noise, then classify with w·f + b."""
    yhat = []
    for x in X:
        f = _features_from_counts(counts_with_noise(fmap, x, shots=shots, p=p, ro_flip=ro_flip), shots)
        yhat.append(1 if (w @ f + b) >= 0 else 0)
    return np.array(yhat, dtype=int)
```

**Methodology Analysis**

**Simulator: AerSimulator(seed_simulator=777) for reproducibility.**

* Gate noise: a 1-qubit depolarizing channel with probability p is attached to single-qubit gates ["id","rz","sx","x"].

* Readout loss: after sampling counts, we flip the LSB of each bitstring with probability ro_flip, modeling asymmetric, lossy readout.

* Features from counts: from 2-qubit bitstrings, compute ⟨Z⊗I⟩ and ⟨I⊗Z⟩ via marginals.

* Mitigation (A only): ZNE with two noise settings, p and 1.5p, then linear extrapolation
f₀ ≈ 2·f(p) − f(1.5p).

* Classifier: both pipelines use the same linear readout w·f + b, isolating the effect of mitigation rather than classifier differences.

* Parameters used in figures: typically p ∈ [0, 0.075], ro_flip = 0.04, shots = 1500.

**Task 6 - Accuracy vs depolarizing noise (QALIS vs CRQC–LLM)**

```python
# === Figure E.2.7a — Accuracy vs depolarizing noise (QALIS vs CRQC–LLM) ===
# Requirements: functions fmap_A, fmap_B, mitigated_pred, unmitigated_pred,
# data X, y, linear readout (w, b), and save_e_figure helper already defined.

ps = np.linspace(0.0, 0.075, 7)   # depolarizing noise levels
ro = 0.04                         # readout flip probability (loss channel)
shots = 1500

acc_A, acc_B = [], []
for p in ps:
    yA = mitigated_pred(fmap_A, X, w, b, shots=shots, p=p, ro_flip=ro)
    yB = unmitigated_pred(fmap_B, X, w, b, shots=shots, p=p, ro_flip=ro)
    acc_A.append(accuracy_score(y, yA))
    acc_B.append(accuracy_score(y, yB))

fig, ax = plt.subplots(figsize=(5.6, 3.6))
ax.plot(ps, acc_A, marker="o", label="QALIS-style (mitigated)")
ax.plot(ps, acc_B, marker="s", label="CRQC–LLM (unmitigated)")
ax.set_xlabel("Depolarizing noise probability p")
ax.set_ylabel("Accuracy")
ax.set_ylim(0.0, 1.0)
ax.grid(True, alpha=0.35)
ax.legend(loc="best", frameon=False)
ax.set_title("Accuracy vs depolarizing noise (fixed readout loss)")

# Save (Appendix E mapping)
save_e_figure("Figure E.2.7a", "P2_AdvLab07_E.2.7a")
```
